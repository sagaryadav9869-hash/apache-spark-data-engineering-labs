```python
# ========================================
# Google Colab: Practicals on RDD with PySpark
# Covers (a) and (b) requirements

#This notebook is a step-by-step demo of PySpark RDD operations:
#Creating RDDs from lists, text, and CSVs.
#Performing actions (count, first, collect).
#Using transformations (map, reduceByKey, groupByKey).
#Managing partitions (repartition, coalesce).
#Using broadcast & accumulators.
#Converting RDDs to DataFrames for easier SQL-like operations.

# ========================================
# 1️⃣ Install and configure PySpark in Colab
!pip install pyspark

# 2️⃣ Import required libraries
from pyspark.sql import SparkSession
from pyspark import SparkContext, SparkConf

# 3️⃣ Create SparkSession & SparkContext
spark = SparkSession.builder \
    .appName("RDD Practicals in PySpark") \
    .getOrCreate()

sc = spark.sparkContext
```

Practicals on RDD (Resilient Distributed Dataset) with Scala Operations and transformations.
a) Use of next mentioned operations and functions: Parallelize, Read text file, Read CSV, Create
RDD, Actions, Pair Functions, Repartition and Coalesce, Shuffle Partitions, Broadcast
Variables, Accumulator Variables and Convert RDD to DataFrame

b) Use of the next mentioned operations : Read multiple text files into RDD, Read CSV file into
RDD, Create an empty RDD, RDD Pair Functions and Generate DataFrame from RDD.

```python
# (a) OPERATIONS AND FUNCTIONS
# ======================================================
# 1. Parallelize
# 2. Read text file
# 3. Read CSV (as RDD)
# 4. Create RDD (from list)
# 5. Actions
# 6. Pair Functions (reduceByKey)
# 7. Repartition & Coalesce
# 8. Shuffle Partitions (with repartition)
# 9. Broadcast Variables
#10. Accumulator Variables
# 11. Convert RDD to DataFrame

import csv
# 1. Parallelize
data = [1, 2, 3, 4, 5]
rdd_parallel = sc.parallelize(data)
print("Parallelized RDD:", rdd_parallel.collect())

# 2. Read text file
# Creating a sample text file
with open("sample_text.txt", "w") as f:
    f.write("Hello Spark\nThis is RDD Practical\nPySpark with Colab")
rdd_text = sc.textFile("sample_text.txt")
print("\nText File RDD:", rdd_text.collect())
```

```python
# 3. Read CSV (as RDD)
import csv
import os

# Create a sample CSV
with open("sample_data.csv", "w") as f:
    f.write("id,name,marks\n1,John,85\n2,Alice,90\n3,Bob,78")
rdd_csv = sc.textFile("sample_data.csv").map(lambda line: line.split(","))
print("\nCSV RDD:", rdd_csv.collect())
```

```python
# 4. Create RDD (from list)
rdd_list = sc.parallelize([("A", 10), ("B", 20), ("C", 30)])
print("\nList RDD:", rdd_list.collect())
```

```python
# 5. Actions
count_val = rdd_list.count()
first_val = rdd_list.first()
print("\nRDD Count:", count_val)
print("First Element:", first_val)
```

```python
# 6. Pair Functions (reduceByKey)
rdd_pairs = sc.parallelize([("math", 80), ("math", 90), ("eng", 70)])
rdd_reduced = rdd_pairs.reduceByKey(lambda a, b: a + b)
print("\nReduced by Key:", rdd_reduced.collect())
```

```python
# 7. Repartition & Coalesce
rdd_repart = rdd_parallel.repartition(4)
rdd_coalesce = rdd_parallel.coalesce(2)
print("\nPartitions after repartition:", rdd_repart.getNumPartitions())
print("Partitions after coalesce:", rdd_coalesce.getNumPartitions())
```

```python
# 8. Shuffle Partitions (with repartition)
rdd_shuffle = rdd_parallel.repartition(3)
print("\nPartitions after shuffle:", rdd_shuffle.getNumPartitions())
```

```python
# 9. Broadcast Variables
broadcast_var = sc.broadcast([1, 2, 3])
print("\nBroadcast Variable:", broadcast_var.value)
```

```python
# 10. Accumulator Variables
accum = sc.accumulator(0)
def add_num(x):
    global accum
    accum += x

rdd_parallel.foreach(add_num)
print("\nAccumulator Value:", accum.value)
```

```python
# 11. Convert RDD to DataFrame
df_from_rdd = spark.createDataFrame(rdd_list, ["Letter", "Value"])
print("\nDataFrame from RDD:")
df_from_rdd.show()
```

```python
# (b) MORE OPERATIONS
# 1. Read multiple text files into RDD
# 2. Read CSV file into RDD (already shown above)
# 3. Create an empty RDD
# 4. RDD Pair Functions Example (mapValues, groupByKey)
# 5. Generate DataFrame from RDD
# ======================================================

# 1. Read multiple text files into RDD
with open("file1.txt", "w") as f:
    f.write("This is file 1")
with open("file2.txt", "w") as f:
    f.write("This is file 2")

rdd_multi_text = sc.textFile("file1.txt,file2.txt")
print("\nMultiple Text Files RDD:", rdd_multi_text.collect())
```

```python
# 2. Read CSV file into RDD (already shown above)
# Let's filter out the header
header = rdd_csv.first()
rdd_csv_noheader = rdd_csv.filter(lambda row: row != header)
print("\nCSV without header:", rdd_csv_noheader.collect())
```

```python
# 3. Create an empty RDD
empty_rdd = sc.emptyRDD()
print("\nEmpty RDD:", empty_rdd.collect())
```

```python
# 4. RDD Pair Functions Example (mapValues, groupByKey)
rdd_pair2 = sc.parallelize([("math", 85), ("math", 95), ("eng", 75)])
grouped_rdd = rdd_pair2.groupByKey().mapValues(list)
print("\nGrouped by Key:", grouped_rdd.collect())
```

```python
# 5. Generate DataFrame from RDD
df_from_pair = spark.createDataFrame(rdd_pair2, ["Subject", "Marks"])
print("\nDataFrame from Pair RDD:")
df_from_pair.show()
```
